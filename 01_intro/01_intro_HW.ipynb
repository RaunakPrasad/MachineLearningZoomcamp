Q7. SUM OF WEIGHTS


```python
asia_cars = df[df['origin'] == 'Asia']
```

```python
asia_cars = asia_cars[['vehicle_weight', 'model_year']]
```

```python
asia_cars = asia_cars.head(7)
```

```python
asia_cars
```

```python
X = asia_cars.values
```

```python
X
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
w.sum()
```

```python
np.sum(w)
```
